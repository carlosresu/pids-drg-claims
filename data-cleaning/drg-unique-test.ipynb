```r
library(data.table)

# Define the file path template (with the space handled correctly in the string)
file_path_template <- "/home/resurreccion_cmc_gmail_com/drg-pipeline/data-cleaning/data/claims/raw/parts/claims_extract_CLAIMS 2018_part_%02d_of_15.rds"

# Initialize an empty data.table
combined_result <- data.table()

# Read all 15 parts
for (i in 1:2) {
  # Construct the file path with leading zeroes
  file_path <- sprintf(file_path_template, i)

  # Read the CSV file and append to the combined_result
  result <- readRDS(file_path)
  # message(nrow(result))
  str(result)
  combined_result <- rbindlist(list(combined_result, result), use.names = TRUE, fill = TRUE)
}

# Calculate total number of rows
total_nrow <- nrow(combined_result)

# Calculate unique number of PSEUDO_CLAIMSERIES
unique_id_series_nrow <- combined_result[, uniqueN((PSEUDO_CLAIMSERIES))]

# Print the results
cat("Total number of rows:", total_nrow, "\n")
cat("Unique number of PSEUDO_CLAIMSERIES:", unique_id_series_nrow, "\n")

# Check if the total number of rows equals the unique PSEUDO_CLAIMSERIES count
if (total_nrow == unique_id_series_nrow) {
  cat("Total number of rows equals the number of unique PSEUDO_CLAIMSERIES.\n")
} else {
  cat("Total number of rows does not equal the number of unique PSEUDO_CLAIMSERIES.\n")
}
```
