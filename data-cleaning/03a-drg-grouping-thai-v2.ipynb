```python
from pathlib import Path
import subprocess
import sys
import os

# Define root directory as current working directory
root_dir = Path.cwd()

# Define relevant directories
chkpt_4_input_dir = root_dir / "data-cleaning" / "data" / "chkpts" / "chkpt_4_thai_master_input"
chkpt_5_output_dir = root_dir / "data-cleaning" / "data" / "chkpts" / "chkpt_5_thai_output"

# Define gsutil input files
input_files = [
    "chkpt_4_thai_grouper_input_2018_full_part_1_of_2.txt",
    "chkpt_4_thai_grouper_input_2018_full_part_2_of_2.txt",
    "chkpt_4_thai_grouper_input_2019_full_part_1_of_2.txt",
    "chkpt_4_thai_grouper_input_2019_full_part_2_of_2.txt",
    "chkpt_4_thai_grouper_input_2020_full_part_1_of_2.txt",
    "chkpt_4_thai_grouper_input_2020_full_part_2_of_2.txt",
    "chkpt_4_thai_grouper_input_2021_full_part_1_of_1.txt",
    "chkpt_4_thai_grouper_input_2022_full_part_1_of_2.txt",
    "chkpt_4_thai_grouper_input_2022_full_part_2_of_2.txt",
    "chkpt_4_thai_grouper_input_2023_full_part_1_of_2.txt",
    "chkpt_4_thai_grouper_input_2023_full_part_2_of_2.txt"
]

bucket_prefix = "gs://pids-drg-data/data/phic/thai/pre"

# Step 1: Check if gcloud CLI is installed
def is_tool_installed(tool_name):
    try:
        subprocess.run([tool_name, "--version"], check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        return True
    except Exception:
        return False

if not is_tool_installed("gcloud"):
    print("❌ Google Cloud CLI (gcloud) is not installed.")
    print("👉 Please copy and paste the following commands into Command Prompt (cmd.exe):\n")
    print(":: Download and install Google Cloud SDK")
    print("curl -o gcloud_installer.exe https://dl.google.com/dl/cloudsdk/channels/rapid/GoogleCloudSDKInstaller.exe")
    print("start gcloud_installer.exe\n")
    print(":: After installation completes, initialize gcloud")
    print("gcloud init\n")
    input("⏸️ Once you've installed and initialized gcloud, press Enter to continue or Ctrl+C to abort...")
else:
    print("✅ gcloud CLI is installed.")
    print("👉 If you haven’t already initialized gcloud, please run the following in Command Prompt:")
    print("gcloud init\n")
    input("⏸️ Press Enter once you've completed 'gcloud init'...")

# Step 2: Ensure input directory exists
chkpt_4_input_dir.mkdir(parents=True, exist_ok=True)

# Step 3: Download input files using gsutil
print("📥 Downloading input files via gsutil...")
gsutil_cp_cmd = ["gsutil", "-m", "cp"]
gsutil_cp_cmd.extend([f"{bucket_prefix}/{fname}" for fname in input_files])
gsutil_cp_cmd.append(str(chkpt_4_input_dir))

subprocess.run(gsutil_cp_cmd, check=True)

# Step 4: Prompt user to run the Thai Batch Grouper
response = input("\n🚨 Run the Thai Batch Grouper manually, then press 'y' to continue: ")
if response.lower() != 'y':
    print("❌ Thai Batch Grouper not confirmed as completed. Aborting.")
    sys.exit(1)

# Step 5: Upload output files back to GCS
print("📤 Uploading Thai Batch Grouper output files to GCS...")
output_upload_cmd = [
    "gsutil", "-m", "cp", "-r", str(chkpt_5_output_dir / "*"),
    "gs://pids-drg-data/data/phic/thai/post/"
]
subprocess.run(" ".join(output_upload_cmd), shell=True, check=True)

print("✅ All steps completed.")
```
