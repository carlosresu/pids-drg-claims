```r
# Define a function to kill all unused R processes
kill_all_r_processes <- function() {
  # Run tasklist to list all processes and search for R processes
  tasklist_output <- system("tasklist /FI \"IMAGENAME eq Rterm.exe\" /FO CSV", intern = TRUE)

  # Check if R processes are running
  if (length(tasklist_output) > 1) {
    # The first line of the output is the header, so we skip it
    r_processes <- tasklist_output[-1]

    # Print the R processes found (for debugging)
    cat("R processes found:\n")
    print(r_processes)

    # Kill all R processes
    system("taskkill /F /IM Rterm.exe", wait = TRUE)

    cat("All R processes have been terminated.\n")
  } else {
    cat("No R processes found.\n")
  }
}

# Call the function
kill_all_r_processes()
```
